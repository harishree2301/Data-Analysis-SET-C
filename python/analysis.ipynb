```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
assessments= pd.read_csv('assessment.csv')
courses= pd.read_csv('courses.csv')
```

```python
print(assessments.dtypes)
print(courses.dtypes)
```

```python
assessments=assessments.drop_duplicates()
```

```python
df=assessments.merge(courses, how='left', on='course_id')

print(df)
```

```python
df["pass_flag"]= (df["score"] >= 60).astype(int)
print(df[["score", "pass_flag"]])
```

```python
department_summary=(df.groupby("department").agg(total_assessments=("assessment_id", "count"),passed_assessments=("pass_flag", "sum")).reset_index())

department_summary["pass_rate"]= department_summary["passed_assessments"]/department_summary["total_assessments"]*100

department_summary["pass_rate"]= department_summary["pass_rate"].round(2)
print(department_summary)
```

```python
month_order=["Jan","Feb","Mar"]

df["month"]= pd.Categorical(df["month"], categories=month_order, ordered=True)

month_avg=(df.groupby("month",observed=False)["score"].mean().round(2))
print(month_avg)

```

```python
plt.title("Average Score by Month")
plt.xlabel("Month")
plt.ylabel("Average Score")
plt.save.fig("pyhton_charts.png")
plt.show()
```

```python
df.to_csv("clean_data.csv",index=False)
department_summary.to_csv("python_summary.csv",index=False)
```
